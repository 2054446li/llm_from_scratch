```python
import torch
import torch.nn as nn
import torch.nn.functional as F
```

FFN(RELU)

```python
class FFN_Relu(nn.Module):
    def __init__(self,dim):
        super(FFN_Relu,self).__init__()
        self.linear1=nn.Linear(dim,4*dim)
        self.linear2=nn.Linear(4*dim,dim)

    def forward(self,x):
        return x+self.linear2(F.relu(self.linear1(x)))
```

FFN (swiglu)

```python
class silu(nn.Module):
    def __init__(self):
        super(silu,self).__init__()
    def forward(self,x):
        return x*torch.sigmoid(x)

class FFN_swiglu(nn.Module):
    def __init__(self,dim):
        super(FFN_swiglu,self).__init__()
        hidden_dim=(int)(dim*(8)//3)
        self.W1=nn.Linear(dim,hidden_dim)
        self.W2=nn.Linear(dim,hidden_dim)
        self.W3=nn.Linear(hidden_dim,dim)
        self.swiglu=silu()

    def forward(self,x):
        return x+self.W3(self.swiglu(self.W2(x))*(self.W1(x)))
```
